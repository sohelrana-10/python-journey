```python
import numpy as np

mat=np.matrix([[1,2,3],
               [4,5,6]])
print(mat)

print()

arr=np.array([[1,2,3],
               [4,5,6]])
print(arr)

print("See the diff with multiplication:")

print("Multiplication of array by 3:",arr*3)

print("Multiplication of matrix by 3:",mat*3)

print("We are taking another array and matrix.")

mat2=np.matrix([[1,2,3],
               [4,5,6]])

arr2=np.array([[1,2,3],
               [4,5,6]])

print()
print("Multiplication of the two array:",arr*arr2)
print("Multiplication of two matrix:",mat.dot(mat2))#will give error bcz not compatible of m x n /= n x m
```

```python
mat3=np.matrix([[1,2],[3,4]])
mat3=np.linalg.matrix_power(mat3,2)
print(mat3)

mat3=np.linalg.matrix_power(mat3,0)
print(mat3)

mat3=np.linalg.matrix_power(mat3,-1)
print(mat3)
```
