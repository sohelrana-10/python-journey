```python
# Broadcasting example using NumPy
import numpy as np

# Create a 2D array with 1 row and 3 columns
arr = np.array([[1, 2, 3]])

# Create another 2D array with 3 rows and 1 column
arr2 = np.array([[1],
                 [2],
                 [3]])

# Print the shapes of both arrays
# arr.shape = (1, 3)
# arr2.shape = (3, 1)
print(arr.shape)
print(arr2.shape)

# Perform element-wise multiplication using broadcasting
# NumPy automatically expands both arrays to match each other’s shape
# arr becomes [[1, 2, 3],
#              [1, 2, 3],
#              [1, 2, 3]]
# arr2 becomes [[1, 1, 1],
#               [2, 2, 2],
#               [3, 3, 3]]
# Result (arr3) = element-wise multiplication of the expanded arrays
arr3 = arr * arr2

# Print the resulting 3x3 array
print(arr3)
```

```python
# Random number generator and shuffle example using NumPy

import numpy as np

# Create a random number generator (RNG) object
# You can pass a seed value like np.random.default_rng(seed=1)
# to get the same random numbers every time you run the code.
rand = np.random.default_rng()

# Generate a 2x2 array of random integers between 1 (inclusive) and 11 (exclusive)
# So the numbers will be from 1 to 10
arr = rand.integers(1, 11, (2, 2))
print("Original array:\n", arr)

# Shuffle the array *in-place* along the first axis (rows)
# It randomly reorders the rows, but does not mix individual elements.
rand.shuffle(arr)
print("\nShuffled array:\n", arr)
```

```python
# Random number generator and shuffle and choice example using NumPy

import numpy as np

# Create a random number generator (RNG) object
# You can pass a seed value like np.random.default_rng(seed=1)
# to get the same random numbers every time you run the code.
rand = np.random.default_rng()

# Generate a 2x2 array of random integers between 1 (inclusive) and 11 (exclusive)
# So the numbers will be from 1 to 10
arr = rand.integers(1, 11, (2, 2))
print("Original array:\n", arr)

# Shuffle the array *in-place* along the first axis (rows)
# It randomly reorders the rows, but does not mix individual elements.
rand.shuffle(arr)
print("\nShuffled array:\n", arr)

choice=rand.choice(arr,2)
print("Random selected choice is:\n",choice)
```
