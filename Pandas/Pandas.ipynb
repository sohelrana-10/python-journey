```python
import pandas as pd

# Data
names = ["Sohel Rana", "Meher", "Afrin", "Queen", "Rahim", "Sara", "Imran", "Nazia", "Tariq", "Rita"]
marks = [100, 70, 90.5, 80, 85, 92, 76, 88, 95, 67]
ages = [20, 19, 21, 22, 20, 19, 23, 21, 22, 20]
cities = ["Dhaka", "Chittagong", "Sylhet", "Dhaka", "Khulna", "Barishal", "Rangpur", "Rajshahi", "Dhaka", "Chittagong"]

# Create Series with marks (labeled by names)
student_marks = pd.Series(marks, index=names)  
# pd.Series(): creates a one-dimensional labeled array

# Create a DataFrame with all details
student_details = pd.DataFrame({
    "Marks": marks,
    "Age": ages,
    "City": cities
}, index=names)  
# pd.DataFrame(): creates a two-dimensional labeled data structure (rows & columns)

# Print examples
print(student_marks)
print(student_details)
```
