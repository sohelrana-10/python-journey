```python
# How to write to a file in write ("w") mode

# Open the file in write mode ("w")
# If the file doesn't exist, Python will create a new one
# If it already exists, its content will be overwritten
f = open("Test.txt", "w")

# Write a string to the file
#f.write("I am writing using the default write() function.")

# Close the file to save changes
f.close()
```

```python
# Opening a file in truncated mode ("w+")

# "w+" mode opens the file for both reading and writing
# BUT it also truncates (clears) the file as soon as it is opened.

# So any existing content will be deleted immediately.

with open("Test.txt", "w+") as f:
    # Trying to read right after opening will print nothing,
    # because the file was truncated (emptied).
    print("Reading immediately after opening (file is empty):")
    print(f.read())  # prints nothing

    # Now, write new content to the file
    f.write("Nothing was printed earlier-now I am stored here!")

    # Move the file pointer to the beginning to read the new content
    f.seek(0)
    print("\nAfter writing new content:")
    print(f.read())
```

```python
# How to write user input to a file in Python

# Take input from the user
data = input("Enter the data to write into the file: ")

# Open the file in write ("w") mode
# If the file doesn't exist, it will be created
# If it exists, its previous content will be replaced
f = open("Test.txt", "w")

# Write the user's input into the file
f.write(data)

# Close the file to ensure data is saved properly
f.close()
```

```python
# How to read a file in read ("r") mode

print("Read mode reads all the content from a file.")

# Open the file in read mode ("r")
# This will throw an error if the file does not exist
f = open("Test.txt", "r")

# Read the entire content of the file
data = f.read()

# Print the content to the console
print(data)

# Close the file after reading
f.close()
```

```python
# If you want to override content from the beginning of the file

# Open the file in read and write mode ("r+")
# This mode allows both reading and writing, starting from the beginning

with open("Test.txt", "r+") as f:
    print("Before overriding:")
    
    # Read and display the current file content
    data = f.read()
    print(data)
    
    # Move the file pointer to the beginning before writing
    f.seek(0)
    f.write("I am overriding from the first!")
    
    # Move the pointer back to the beginning to read the updated content
    f.seek(0)
    print("\nAfter overriding:")
    data = f.read()
    print(data)
```

```python
# Opening a file in append and read mode ("a+")

# "a+" mode opens the file for both appending and reading.
# The file pointer is placed at the END of the file, so reading right away returns nothing.
# You must move the pointer to the beginning to read existing content.

with open("Test.txt", "a+") as f:
    # Move the file pointer to the beginning before reading
    f.seek(0)
    data = f.read()
    print("Before appending:")
    print(data)

    # Write (append) new data at the end of the file
    f.write("\nI am appended data.")  # added newline for readability

    # Move back to the beginning to show updated content
    f.seek(0)
    print("\nAfter appending:")
    print(f.read())
```

```python
class FileNotFound(Exception):
    pass

try:
    filename = "tes.axt"
    if not filename.endswith(".txt"):
        raise FileNotFound("Only .txt files are allowed")

    with open(filename) as f:
        for line in f:
            print(line.strip())

except FileNotFound as e:
    print("Custom Exception:", e)

except FileNotFoundError as e:
    print("Python Built-in Exception:", e)

except Exception as obj:
    print("Other Exception:", obj)

print("Rest of the code")
```

```python
def count_words(filename):
    """Count the approximate number of words in a file."""
    try:
        with open(filename) as f_obj:
            contents = f_obj.read()
    except FileNotFoundError:
        msg = "Sorry, the file " + filename + " does not exist."
        print(msg)
    else:
        # Count approximate number of words in the file.
        words = contents.split()
        num_words = len(words)
        print("The file " + filename + " has about " + str(num_words) + " words.")


filenames = ['alice.txt', 'siddhartha.txt', 'moby_dick.txt', 'little_women.txt']
for filename in filenames:
    count_words(filename)
```
